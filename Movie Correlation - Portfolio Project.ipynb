```python
# We need to see if we have any missing data
# Let's loop through the data and see if there is anything missing


#In pandas, use .isna() or .isnull() to check for missing data—they do the same thing.
#Watch out: Empty text (''), spaces ('   '), and the text 'nan' are not considered missing by .isna().


for col in df.columns:  #Goes through each column name, one at a time.
     pct_missing= np.mean(df[col].isnull()) # pandas calculates the average# pandas calculates the average  
                  #np.mean(...): Treats True as 1 and False as 0, then calculates the average.
                  #df[col].isnull() :Checks every value in that column: missing = True, present = False.
     print('{} - {}%'.format(col, round(pct_missing*100.0)))
```

```python
## NumPy calculates the average
for col in df.columns:
    pct_missing = df[col].isnull().mean()
    print(f'{col} - {pct_missing:.1%}')
```
